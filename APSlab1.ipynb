```python
"""Multiplication rule for independent events

A coin is tossed and a die is rolled 10,000 times.

Define:

obtaining Head

obtaining an even number

Verify: P(A∩B)=P(A)P(B) """

import random
Roll=10000
heads=0
even=0
both=0
for i in range(Roll):
  coin_result=random.choice(["Head" ,"Tail"])
  dice_result=random.randint(1,6)
  if coin_result == "Head":
    heads = heads + 1
  if dice_result % 2 == 0:
    even = even + 1
  if coin_result == "Head" and dice_result % 2 == 0:
    both = both + 1
Prob_A=heads/Roll
Prob_B=even/Roll
Prob_AB=both/Roll

print("Probability of obtaining head= " ,Prob_A)
print("Probability of obtaining even number= " ,Prob_B)
print("Probability of obtaining both= ", Prob_AB)
print("P(A) * P(B) :", Prob_A * Prob_B)






```
