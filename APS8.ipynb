```python

import numpy as np

# Step 1: Generate 10,000 random values from normal distribution
population = np.random.normal(loc=50, scale=10, size=10000)

# Step 2: Calculate population mean
pop_mean = np.mean(population)

# Step 3: Calculate population standard deviation
pop_std = np.std(population)

# Step 4: Randomly select 30 samples
sample = np.random.choice(population, size=30, replace=False)

# Step 5: Calculate sample mean
sample_mean = np.mean(sample)

# Step 6: Calculate sample standard deviation
sample_std = np.std(sample, ddof=1)

# Step 7: Calculate standard error
standard_error = sample_std / np.sqrt(30)

# Display results
print("Population Mean:", pop_mean)
print("Population Standard Deviation:", pop_std)

print("\nSample Mean:", sample_mean)
print("Sample Standard Deviation:", sample_std)
print("Standard Error:", standard_error)
```
