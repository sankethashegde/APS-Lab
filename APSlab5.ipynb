```python
#PROBLEM 1
import numpy as np

# Set seed for reproducibility
np.random.seed(42)

p = 0.30
sample_size = 20

# Generate 20 Bernoulli observations
samples = np.random.binomial(n=1, p=p, size=sample_size)
print("Generated Samples:", samples)
```

```python
#clickcounts
click_cont = np.sum(samples == 1)
nonclick_cont = np.sum(samples == 0)

print("Number of clicks:", click_cont)
print("Number of non-clicks:", nonclick_cont)
```

```python
#emperical pmf
empirical_p_0 = nonclick_cont / sample_size
empirical_p_1 = click_cont / sample_size

print(f"Empirical P(X=0): {empirical_p_0:.2f}")
print(f"Empirical P(X=1): {empirical_p_1:.2f}")
```

```python
import matplotlib.pyplot as plt

outcomes = [0, 1]
theoretical_pmf = [0.70, 0.30]
empirical_pmf = [empirical_p_0, empirical_p_1]

x = np.arange(len(outcomes))
width = 0.35

plt.figure(figsize=(7, 5))
plt.bar(x - width/2, theoretical_pmf, width, label='Theoretical PMF', color='skyblue')
plt.bar(x + width/2, empirical_pmf, width, label='Empirical PMF (N=20)', color='orange')

plt.xlabel('Outcome (X)')
plt.ylabel('Probability')
plt.title('Comparison of Theoretical and Empirical PMF')
plt.xticks(x, ['0 (No Click)', '1 (Click)'])
plt.ylim(0, 1.0)
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
#PROBLEM 2
import numpy as np

# Parameters
n_transactions = 50
p_flag = 0.04
n_simulations = 20000
max_capacity = 4

# Simulate 20,000 standard 1-hour periods
np.random.seed(42)
flagged_counts = np.random.binomial(n=n_transactions, p=p_flag, size=n_simulations)

# Analytical & Simulated Metrics
mean_flagged = np.mean(flagged_counts)
theoretical_mean = n_transactions * p_flag

# Capacity exceedance (hours with > 4 flagged transactions)
over_capacity_hours = np.sum(flagged_counts > max_capacity)
prob_over_capacity = np.mean(flagged_counts > max_capacity)

# Unexamined flagged transactions
unexamined = np.maximum(0, flagged_counts - max_capacity)
total_unexamined = np.sum(unexamined)
avg_unexamined_per_hour = np.mean(unexamined)

print(f"Simulated Mean Flagged Transactions per Hour: {mean_flagged:.4f} (Theoretical: {theoretical_mean:.2f})")
print(f"Hours Exceeding Capacity (> 4 flagged): {over_capacity_hours} / {n_simulations} ({prob_over_capacity * 100:.2f}%)")
print(f"Total Unexamined Flagged Transactions across 20,000 hours: {total_unexamined}")
print(f"Average Unexamined Flagged Transactions per hour: {avg_unexamined_per_hour:.4f}")
```
